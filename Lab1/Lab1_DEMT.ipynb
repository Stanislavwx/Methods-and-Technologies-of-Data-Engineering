## Зчитування та розвідка (Завдання 1-3)
Перший блок завантажує дані та виводить базову інформацію для перших пунктів звіту.

```python
import pandas as pd

# 1. Зчитування даних
df = pd.read_csv('sales__U.csv', encoding='windows-1251', sep=';')

# 2. Опис даних (скопіюй цей текст у звіт):
print("Дані містять інформацію про транзакції: продані товари, категорії, ціни, дати та демографію клієнтів (вік, стать, штат, регіон).")

# 3. Розвідка даних
print("\nКолонки:", df.columns.tolist())
df.info()
```

## Очищення даних (Трансформація)
Замінюємо коми на крапки та перетворюємо текст на числа


```python
df['total'] = pd.to_numeric(df['total'].astype(str).str.replace(',', '.'), errors='coerce')
df['qty_ordered'] = pd.to_numeric(df['qty_ordered'].astype(str).str.replace(',', '.'), errors='coerce')

print("Трансформацію завершено: колонки з грошима та кількістю готові до аналізу.")
```

## Відповіді на 5 простих питань

```python
print("1. Який загальний виторг магазину?")
print(df['total'].sum())
# Пояснення: Базова фінансова метрика загальних надходжень.

print("\n2. Скільки всього товарів було продано?")
print(df['qty_ordered'].sum())
# Пояснення: Показує фізичний обсяг реалізованої продукції.

print("\n3. Які 3 категорії товарів принесли найбільше доходу?")
print(df.groupby('category')['total'].sum().nlargest(3))
# Пояснення: Визначає ключові драйвери прибутку компанії.

print("\n4. Яка стать клієнтів робить більше покупок (за кількістю чеків)?")
print(df['Gender'].value_counts())
# Пояснення: Допомагає зрозуміти основну цільову аудиторію.

print("\n5. (Робота з датами) Який дохід був у кожному році?")
print(df.groupby('year')['total'].sum())
# Пояснення: Оцінка глобальної динаміки бізнесу з року в рік.
```

## Відповіді на 5 складних питань

```python
print("6. (Робота з датами) У якому місяці заробили найбільше грошей?")
print(df.groupby('month')['total'].sum().sort_values(ascending=False).head(1))
# Пояснення: Виявляє найуспішніший сезон або вплив акцій.

print("\n7. Який середній вік клієнтів у розрізі категорій?")
print(df.groupby('category')['age'].mean().round(1))
# Пояснення: Допомагає налаштувати таргетовану рекламу певних товарів.

print("\n8. Які регіони є найприбутковішими?")
print(df.groupby('Region')['total'].sum().sort_values(ascending=False))
# Пояснення: Аналіз ефективності локальних ринків.

print("\n9. Яка середня сума однієї транзакції (середній чек)?")
print(round(df['total'].mean(), 2))
# Пояснення: Ключовий показник купівельної спроможності аудиторії.

print("\n10. Зведена таблиця: сумарний дохід по регіонах (рядки) та категоріях (стовпці)")
report = pd.pivot_table(df, values='total', index='Region', columns='category', aggfunc='sum', fill_value=0)
display(report)
# Пояснення: Готова вітрина даних (Data Mart) для побудови дашбордів.
```
